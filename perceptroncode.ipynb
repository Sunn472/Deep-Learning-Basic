**Import all the required frameworks**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_classification

x,y=make_classification(n_samples = 100, n_features=2, n_redundant=0, n_informative=2, n_clusters_per_class=1, class_sep=0.3, random_state=1)
```

```python
x
```

```python
y
```

```python
plt.scatter(x[y==0][:,0],x[y==0][:,1],color='red',label='Class 0')
plt.scatter(x[y==1][:,0],x[y==1][:,1],color='blue',label='Class 0')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Linearly Separable Data')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.linear_model import Perceptron
p=Perceptron()
p.fit(x,y)
```

```python
p.coef_
```

```python
p.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(x,y,clf=p,legend=2)
```

```python
x=np.array([[0,0],[0,1],[1,0],[1,1]])
y=np.array([0,0,0,1])

model=Perceptron()
model.fit(x,y)

plot_decision_regions(x,y,clf=model,legend=2)
plt.title('AND Gate')
plt.show()
```

```python
#above class 1 and class 0 are separable
y=np.array([0,1,1,1])

model=Perceptron()
model.fit(x,y)

plot_decision_regions(x,y,clf=model,legend=2)
plt.title("OR Gate")
plt.show()
```

```python
y=np.array([0,1,1,0])

model=Perceptron()
model.fit(x,y)

plot_decision_regions(x,y,clf=model,legend=2)
plt.title("XOR Gate")
plt.show()
```

**Perceptron doesnt work with non linear data**

**Create the dataframe of the three specific gates**

```python
or_data = pd.DataFrame()
and_data = pd.DataFrame()
xor_data = pd.DataFrame()
```

```python
or_data['input1'] = [1,1,0,0]
or_data['input2'] = [1,0,1,0]
or_data['output'] = [1,1,1,0]
print(or_data)
```

```python
and_data['input1'] = [1,1,0,0]
and_data['input2'] = [1,0,1,0]
and_data['output'] = [1,0,0,0]
print(and_data)
```

```python
xor_data['input1'] = [1,1,0,0]
xor_data['input2'] = [1,0,1,0]
xor_data['output'] = [0,1,1,0]
print(xor_data)
```

```python
clf_xor.fit(xor_data.iloc[:,0:2].values,xor_data.iloc[:,-1].values)
```

**Limitation Of Perceptron: It is easily able to create the linear decision boundaries but for non-linear functions, it won't be able to create the decision boundaries.**

```python
from mlxtend.plotting import plot_decision_regions

# Set up subplots for side-by-side plots
fig,axes = plt.subplots(1,3,figsize=(18,5))

# AND Classifier Decision Boundary
axes[0].set_title('AND Gate Decision Boundary')
plot_decision_regions(X=and_data.iloc[:,0:2].values,y=and_data.iloc[:,-1].values,clf=clf_and,legend=2,ax=axes[0])
axes[0].set_xlabel('Input 1')
axes[0].set_ylabel('Input 2')

# OR Classifier Decision Boundary
axes[1].set_title('OR Gate Decision Boundary')
plot_decision_regions(X=or_data.iloc[:,0:2].values,y=or_data.iloc[:,-1].values,clf=clf_or,legend=2,ax=axes[1])
axes[1].set_xlabel('Input 1')
axes[1].set_ylabel('Input 2')

# AND Classifier Decision Boundary
axes[2].set_title('XOR Gate Decision Boundary')
plot_decision_regions(X=xor_data.iloc[:,0:2].values,y=xor_data.iloc[:,-1].values,clf=clf_xor,legend=2,ax=axes[2])
axes[2].set_xlabel('Input 1')
axes[2].set_ylabel('Input 2')
```
